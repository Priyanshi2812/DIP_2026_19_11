```python
import pandas as pd
from pathlib import Path

folder = Path("/content")

# Load the two country datasets
argentina = pd.read_csv(folder / "Argentina.csv")
brazil = pd.read_csv(folder / "Brazil.csv")

datasets = {
    "Argentina": argentina,
    "Brazil": brazil
}

for country, data in datasets.items():
    print("\n" + "=" * 40)
    print(country)
    print("Rows:", len(data))
    print("Columns:", data.columns.tolist())

    if "field_shared_name" in data.columns:
        print(
            "Unique fields:",
            data["field_shared_name"].nunique()
        )
        print(
            "Missing field IDs:",
            data["field_shared_name"].isna().sum()
        )

    if "crop" in data.columns:
        print("\nCrop counts:")
        print(data["crop"].value_counts(dropna=False))

    if "date" in data.columns:
        dates = pd.to_datetime(
            data["date"],
            dayfirst=True,
            errors="coerce"
        )

        print("\nInvalid dates:", dates.isna().sum())
        print("Earliest date:", dates.min())
        print("Latest date:", dates.max())

    if all(
        col in data.columns
        for col in ["coord_x", "coord_y", "coord_z"]
    ):
        print("\nMissing coordinate values:")
        print(
            data[["coord_x", "coord_y", "coord_z"]]
            .isna()
            .sum()
        )
```

```python
from pathlib import Path
import pandas as pd

output_folder = Path("/content/alignment_check")
output_folder.mkdir(parents=True, exist_ok=True)

report = []

for country, data in datasets.items():

    # Parse dates using DD-MM-YYYY convention
    dates = pd.to_datetime(
        data["date"],
        dayfirst=True,
        errors="coerce"
    )

    report.append({
        "country": country,
        "total_rows": len(data),
        "unique_fields": data["field_shared_name"].nunique(),
        "missing_field_ids": data["field_shared_name"].isna().sum(),
        "missing_dates": dates.isna().sum(),
        "earliest_date": dates.min(),
        "latest_date": dates.max(),
        "missing_coord_x": data["coord_x"].isna().sum(),
        "missing_coord_y": data["coord_y"].isna().sum(),
        "missing_coord_z": data["coord_z"].isna().sum()
    })

report_df = pd.DataFrame(report)

report_path = output_folder / "preliminary_alignment_report.csv"
report_df.to_csv(report_path, index=False)

display(report_df)

print("Saved preliminary alignment report to:", report_path)

```
