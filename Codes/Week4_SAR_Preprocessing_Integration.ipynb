```python
import pandas as pd
import numpy as np

optical = pd.read_csv("Optical_Features.csv")
sar = pd.read_csv("SAR_VV_VH_verified_pixels.csv")

keys = ["country", "farm_identifier", "field_shared_name"]

for df in [optical, sar]:
    df["country"] = df["country"].astype(str).str.strip()
    df["farm_identifier"] = pd.to_numeric(
        df["farm_identifier"], errors="raise"
    ).astype(int)
    df["field_shared_name"] = pd.to_numeric(
        df["field_shared_name"], errors="raise"
    ).astype(int)

sar_field = (
    sar.groupby(keys)
    .agg(
        VV_mean=("VV", "mean"),
        VV_std=("VV", "std"),
        VH_mean=("VH", "mean"),
        VH_std=("VH", "std"),
        SAR_pixel_count=("VV", "count")
    )
    .reset_index()
)

merged = optical.merge(
    sar_field,
    on=keys,
    how="left",
    validate="many_to_one",
    indicator=True
)

print("Optical rows:", len(optical))
print("Unique optical fields:", optical[keys].drop_duplicates().shape[0])
print("Unique SAR fields:", len(sar_field))
print("Optical rows with SAR:", (merged["_merge"] == "both").sum())
print("Optical rows without SAR:", (merged["_merge"] == "left_only").sum())

print("\nMatched fields:")
print(
    merged.loc[merged["_merge"] == "both", keys]
    .drop_duplicates()
    .to_string(index=False)
)

print("\nUnmatched optical fields:")
print(
    merged.loc[merged["_merge"] == "left_only", keys]
    .drop_duplicates()
    .to_string(index=False)
)

merged.drop(columns="_merge").to_csv(
    "Week4_Optical_SAR_Merged.csv", index=False
)

sar_field.to_csv("SAR_field_level_features.csv", index=False)

print("\nFiles saved successfully.")
```

```python
print("Merged shape:", merged.shape)
print("\nMissing values in SAR features:")
print(merged[["VV_mean", "VH_mean"]].isna().sum())

print("\nCrop distribution:")
print(merged["crop"].value_counts())

display(merged.head())
```

```python
summary = pd.DataFrame({
    "Metric": [
        "Optical observations",
        "Unique optical fields",
        "SAR pixel samples",
        "Unique SAR fields",
        "Matched optical observations",
        "Unmatched optical observations",
        "Missing VV values after merge",
        "Missing VH values after merge"
    ],
    "Value": [
        len(optical),
        optical[keys].drop_duplicates().shape[0],
        len(sar),
        len(sar_field),
        (merged["_merge"] == "both").sum(),
        (merged["_merge"] == "left_only").sum(),
        merged["VV_mean"].isna().sum(),
        merged["VH_mean"].isna().sum()
    ]
})

display(summary)
summary.to_csv("Week4_SAR_alignment_summary.csv", index=False)
```
